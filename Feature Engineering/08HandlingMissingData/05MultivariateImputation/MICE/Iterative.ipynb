```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
pd.read_csv("50_Startups.csv").head()
```

```python
df=np.round(pd.read_csv("50_Startups.csv",usecols=['R&D Spend','Administration','Marketing Spend','Profit'])/10000)
```

```python
np.random.seed(10)  # it is use to produce same sample data having seed value 10
df.sample(5)
```

```python
df=df.head(5)
df=df.iloc[:,0:3]
df
```

```python
df.iloc[1,0]=np.nan
df.iloc[3,1]=np.nan
df.iloc[-1,-1]=np.nan
```

```python
df.head()
```

```python
# Step 1 - Impute all missing values with mean of respective col

df0=pd.DataFrame()

df0=df.copy()
df0
```

```python
nan_postions_rows,nan_postions_columns=np.where(df0.isnull())
nan_postions_rows
```

```python
# df0['R&D Spend'].fillna(df0['R&D Spend'].mean(),inplace=True)

df0.fillna(df0.mean(),inplace=True)
```

```python
df0
```

```python
df1=df0.copy()
df1.iloc[nan_postions_rows[0],nan_postions_columns[0]]=np.nan
df1
```

```python
# Use first 3 rows to build a model and use the last for prediction

# X=df.iloc[!df]
not_nan_rows = df1[df1.notna().all(axis=1)].index.tolist()

X=df1.iloc[not_nan_rows,1:3]
X
```

```python
y=df1.iloc[not_nan_rows,0]
y
```

```python
lr=LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[nan_postions_rows[0],1:].values.reshape(1,2))
```
