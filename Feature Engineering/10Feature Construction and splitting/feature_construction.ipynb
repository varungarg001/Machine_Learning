```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

import seaborn as sns
```

```python
df=pd.read_csv("titanic_real.csv",usecols=['Age','Pclass','SibSp','Parch','Survived'])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X=df.drop(columns='Survived')
y=df['Survived']
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

## Apply Feature Construction

```python
X['Family_Size']=X['Parch']+X['SibSp']+1
```

```python
X.head()
```

```python
# df_new = df.copy()

X['Family_Type'] = np.where(
    X['Family_Size']==1,
    0, # 0, alone
    np.where(
        (X['Family_Size']>1) & (X['Family_Size']<=4),
        1,  # small family
        2   # large family
    )
)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_Size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```
