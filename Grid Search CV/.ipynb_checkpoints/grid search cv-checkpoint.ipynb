```python
import pandas as pd
from sklearn import datasets
from sklearn.ensemble import RandomForestClassifier
```

```python
iris=datasets.load_iris()
iris
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
```

```python
df['flower'] = iris.target
```

```python
df['flower'] = df['flower'].apply(lambda x: iris.target_names[x] )
```

```python
df['flower'].value_counts()
```

```python
from sklearn.model_selection import train_test_split

X=df.drop(columns=['flower'])
y=df['flower']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3)
```

```python
X_train.head()
```

```python
from sklearn.svm import SVC
```

```python
model = SVC(kernel='rbf',C=30,gamma='auto')
model.fit(X_train,y_train)
model.score(X_test,y_test)
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np
```

```python
np.mean(cross_val_score(SVC(kernel='linear',C=10,gamma='auto'),X,y,cv=5))
```

```python
np.mean(cross_val_score(SVC(kernel='rbf',C=20,gamma='auto'),X,y,cv=5))
```

```python
np.mean(cross_val_score(SVC(kernel='rbf',C=10,gamma='auto'),X,y,cv=5))
```

```python
kernels = ['rbf','linear']
C=[1,10,20,4,5,14,20,89,70,32,65]
avg_score = {}

for Kval in kernels:
    for cval in C:
        cv_scores = cross_val_score(SVC(kernel=Kval,C=cval,gamma='auto'),X,y,cv=10)
        avg_score[Kval + ' - '+ str(cval)]=np.mean(cv_scores)

avg_score
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
clf = GridSearchCV(SVC(gamma='auto'),{
    'C':[x for x in range(1,51)],
    'kernel':['rbf','linear']
},cv=10,return_train_score=False)

clf.fit(X,y)
# clf.cv_results_
```

```python
p=pd.DataFrame(clf.cv_results_)
```

```python
p.head(1)
```

```python
keyValue={}

for i in p['params'].values:
    print( str(i) + str(p[p['params']==i]['mean_test_score']))
```

```python
p[['params','mean_test_score']]
```

```python
clf.best_params_
```

```python
clf.best_score_
```

```python
# The problem with the gridSearch Cv is resource cost now our dataset is small but it is large it checks for each value of paramtere so tackle this we have the randomized grid search cv
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
rs = RandomizedSearchCV(SVC(gamma='auto'),{
    'C':[x for x in range(1,50)],
    'kernel':['rbf','linear']
},n_iter=5,cv=5)

rs.fit(X,y)
```

```python
pd.DataFrame(rs.cv_results_)[['params','mean_test_score']]
```

```python
rs.best_params_
```

```python
rs.best_score_
```

```python
models = {
    'SVM':{
        'model':SVC(gamma='auto'),
        'params':{
            'C':[1,15,10,20],
            'kernel':['rbf','linear']
        }
    },
    'random_forest':{
        'model':RandomForestClassifier(),
        'params':{
            'n_estimators':[1,40,10,20]
        }
    }
}
```

```python
scores=[]

for model_name,mp in models.items():
    clf=GridSearchCV(mp['model'],mp['params'],cv=10)
    clf.fit(X,y)
    scores.append({
        'model':model_name,
        'best_params':clf.best_params_,
        'best_score':clf.best_score_
    })
```

```python
pd.DataFrame(scores)
```
